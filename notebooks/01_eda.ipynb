```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 200)
sns.set_style("whitegrid")
```

```python
DATA_PATH  = "D:/code/credit_risk_ml_system/data/raw/home_credit/application_train.csv"
df = pd.read_csv(DATA_PATH)
print("Shape:",df.shape)
df.head()
```

```python
df.describe()
df.info()
```

```python
target_dist  = df["TARGET"].value_counts(normalize=True)
print(target_dist)
```

```python
sns.countplot(x=df["TARGET"])
plt.title("Target distribution (default vs non default)")
plt.show()
```

```python
missing_pct = df.isnull().mean().sort_values(ascending=False)
missing_pct.head(10)
```

```python
missing_summary = pd.DataFrame({

    "missing_pct" : missing_pct,
    "missing_count" : df.isnull().sum()
})

missing_summary.head(20)
```

```python
df["DAYS_BIRTH"].describe()
```

```python
df["AGE_YEARS"] = -df["DAYS_BIRTH"]/365
```

```python
sns.boxplot(x="TARGET", y="AGE_YEARS", data=df)

plt.show()
```

```python
df["CREDIT_TO_INCOME"] = df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]

sns.boxplot(x="TARGET", y="CREDIT_TO_INCOME", data=df)
plt.yscale("log")
plt.show()
```

```python
df["ANNUITY_TO_INCOME"] = df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]

sns.boxplot(x="TARGET", y="ANNUITY_TO_INCOME", data=df)
plt.yscale("log")
plt.show()
```
